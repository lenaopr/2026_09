```python
# LMS UU by Search
# No. of Keyword Search Landmark = No. of Search Tips Landmark + No. of Advance Search Landmark = quick + adv
# No. of Search Tips Landmark (quick)

sql = r"""
with temp as 
    (select deviceid,lower(eventlabel) as screen,
     lower(LAG(eventlabel) OVER(ORDER BY sessionid, time)) AS prev_screen
     from `openrice-production.ORGA.SV_202609*`
     where lower(eventlabel) like  'hk.lms2.%.tab.-%' or lower(eventlabel) =  'hk.qcksearch' or lower(eventlabel) like  'hk.advsearch' )

  
select sum(case when prev_screen = 'hk.qcksearch' then 1 end) as quick_,
sum(case when prev_screen = 'hk.advsearch' then 1 end ) as adv_,
count(distinct deviceid) as uu,screen from temp
where (screen = 'hk.lms2.42.tab.-990.1'
or screen = 'hk.lms2.65.tab.-990.1'
or screen = 'hk.lms2.35215.tab.-990.1'
or screen = 'hk.lms2.35329.tab.-990.1')
and prev_screen like '%search%'
group by screen
order by 2 desc

"""
```
